Zip the results.

```python
from pathlib import Path
from zipfile import ZIP_DEFLATED, ZipFile

output_dir = Path("/HHC/outputs")
archive = Path("/content/outputs.zip")

with ZipFile(archive, "w", ZIP_DEFLATED) as bundle:
    for path in output_dir.rglob("*"):
        if path.is_file() and path.suffix in {".csv", ".json", ".jsonl"}:
            bundle.write(path, path.relative_to(output_dir))

print(f"Archive: {archive} ({archive.stat().st_size / 1024**2:.2f} MB)")
```
